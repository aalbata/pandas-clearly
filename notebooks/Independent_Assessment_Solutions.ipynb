## A03. Clean with an audit trail (5 points)

Start with raw_cafe_orders(). Each Order_ID should identify one complete order. Remove exact duplicate rows. Strip Item text and use title case. Standardize Campus to RAK or Dubai, and treat blank Campus as Unknown while retaining a Campus_was_missing flag.
```python
# Complete the task above. What result do you expect?
df = raw_cafe_orders().drop_duplicates().reset_index(drop=True)
df["Item"] = df["Item"].astype("string").str.strip().str.title()
campus_text = df["Campus"].astype("string").str.strip().replace("", pd.NA)
df["Campus_was_missing"] = campus_text.isna()
df["Campus"] = campus_text.str.lower().map({"rak": "RAK", "dubai": "Dubai"}).fillna("Unknown")
df["Quantity_raw"] = df["Quantity"]
df["Unit_Price_raw"] = df["Unit_Price"]
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Unit_Price"] = pd.to_numeric(df["Unit_Price"], errors="coerce")
df["Quantity_unparsed"] = df["Quantity"].isna()
df["Quantity_not_positive_integer"] = df["Quantity"].notna() & ((df["Quantity"] <= 0) | (df["Quantity"] % 1 != 0))
df["Price_missing_or_invalid"] = df["Unit_Price"].isna() | (df["Unit_Price"] <= 0)
bad = df[["Quantity_unparsed", "Quantity_not_positive_integer", "Price_missing_or_invalid"]].any(axis=1)
audit = df.loc[bad, ["Order_ID", "Quantity_raw", "Unit_Price_raw", "Quantity_unparsed", "Quantity_not_positive_integer", "Price_missing_or_invalid"]].copy()
result = df.loc[~bad].copy()
print("Accepted orders:")
print(result[["Order_ID", "Item", "Campus", "Campus_was_missing", "Quantity", "Unit_Price"]])
print("Excluded orders and reasons:")
print(audit)
```
